# Q1) The Probability of Shooting at a Target

A point is chosen at random from a disk of radius $10$. Let $A$ be the event that the point lies within $1$ unit of the boundary.

1. Model the experiment assuming the point is uniformly distributed with respect to area, and compute $P(A)$.
2. Model the experiment assuming the distance to the point from the center is chosen uniformly from $[0,10]$, and the direction is chosen independently and uniformly from $[0,2\pi)$ and compute $P(A)$.
3. Explain why the two answers differ.


# Q1) Probability of a Random Point Near the Boundary

Consider the disk
$$
\Omega = \{(x,y) \in \mathbb{R}^2 : x^2 + y^2 \le 100\}.
$$

Let $A$ be the event that a randomly chosen point lies within $1$ unit of the boundary.  
This corresponds to points whose distance from the center satisfies
$$
9 \le \sqrt{x^2 + y^2} \le 10.
$$

Thus, $A$ represents the outer ring (annulus) of the disk.

---

## 1. Uniform Distribution with Respect to Area

In this model, probability is proportional to area.

- Total area of the disk:
$$
\pi (10)^2 = 100\pi
$$

- Area of region $A$:
$$
\pi(10)^2 - \pi(9)^2 = 100\pi - 81\pi = 19\pi
$$

- Probability:
$$
P(A) = \frac{19\pi}{100\pi} = \frac{19}{100}
$$

\[
\boxed{P(A) = \frac{19}{100}}
\]

---

## 2. Uniform Distribution of the Radius

Here:
- Radius $r \sim \text{Uniform}(0,10)$
- Angle $\theta \sim \text{Uniform}(0, 2\pi)$ (independent)

Since the event depends only on $r$, we consider:

- Total interval length: $10$
- Favorable interval: $[9,10]$ with length $1$

Thus,
$$
P(A) = \frac{1}{10}
$$

\[
\boxed{P(A) = \frac{1}{10}}
\]

---

## 3. Reason for the Difference

The results differ because the two models define "randomness" differently:

- **Area-uniform model**: Every region with equal area is equally likely.  
  Since outer regions occupy more area, points near the boundary are more likely.

- **Radius-uniform model**: Each value of $r$ is equally likely, regardless of how much area it represents.  
  This ignores the fact that larger radii correspond to larger circular regions.

Hence,
$$
\frac{19}{100} \neq \frac{1}{10}
$$

---

## Key Insight

The probability of an event depends not only on the set $\Omega$, but also on the **probability measure** used.

- Uniform over area $\rightarrow$ physically realistic (e.g., dart throwing)
- Uniform over radius $\rightarrow$ artificial sampling process

---

## Interpretation

In real-world situations (like randomly throwing darts at a board), the correct model is the **area-uniform distribution**, giving:
$$
P(A) = \frac{19}{100}.
$$

# Q2) Bertrand's Paradox

Consider the unit circle
$$
x^2 + y^2 = 1.
$$

We want the probability that a randomly chosen chord has length greater than the side length of the inscribed equilateral triangle.

---

## Key Geometric Result

The side length of the inscribed equilateral triangle is
$$
\sqrt{3}.
$$

If a chord is at perpendicular distance $r$ from the center, its length is
$$
L(r) = 2\sqrt{1 - r^2}.
$$

We require
$$
L(r) > \sqrt{3}.
$$

This gives
$$
2\sqrt{1 - r^2} > \sqrt{3}
\;\;\Longrightarrow\;\;
r < \frac{1}{2}.
$$

So the event of interest is:
$$
A = \{ \text{distance from center} < 1/2 \}.
$$

---

## Model 1: Uniform Endpoints (Angle Model)

Choose two endpoints independently and uniformly on the circumference.

By symmetry, fix one endpoint and let $\theta \in [0,\pi]$ be the subtended angle.

Chord length:
$$
L(\theta) = 2\sin(\theta/2).
$$

Condition:
$$
2\sin(\theta/2) > \sqrt{3}
\;\;\Longrightarrow\;\;
\theta > \frac{2\pi}{3}.
$$

Thus,
$$
A_1 = \left(\frac{2\pi}{3}, \pi \right].
$$

Probability:
$$
P_1(A_1) = \frac{\pi - 2\pi/3}{\pi} = \frac{1}{3}.
$$

\[
\boxed{P_1 = \frac{1}{3}}
\]

---

## Model 2: Uniform Distance from Center

Choose a radius and then pick a point uniformly along it.  
Draw the chord perpendicular to the radius at that point.

Here $r \sim \text{Uniform}[0,1]$.

Condition:
$$
r < \frac{1}{2}.
$$

Probability:
$$
P_2 = \frac{1}{2}.
$$

\[
\boxed{P_2 = \frac{1}{2}}
\]

---

## Model 3: Uniform Midpoint in the Disk

Choose the midpoint of the chord uniformly in the unit disk.

The chord is long when the midpoint satisfies:
$$
x^2 + y^2 < \frac{1}{4}.
$$

This is a disk of radius $1/2$.

Probability:
$$
P_3 = \frac{\pi (1/2)^2}{\pi} = \frac{1}{4}.
$$

\[
\boxed{P_3 = \frac{1}{4}}
\]

---

## Final Comparison

| Model | Method | Probability |
|------|--------|------------|
| 1 | Uniform angle/endpoints | $\frac{1}{3}$ |
| 2 | Uniform distance $r$ | $\frac{1}{2}$ |
| 3 | Uniform midpoint (area) | $\frac{1}{4}$ |

---

## Conclusion

Different reasonable interpretations of “random chord” produce different probability measures, leading to different answers.

This is **Bertrand’s paradox**:  
the phrase “chosen at random” is ambiguous unless the sampling method is clearly defined.

In probability theory, specifying the **probability space** is essential to obtain a unique result.

#Q8) Probability of Being a Girl?

A family has two children. Assume each child is equally likely to be a boy or a girl, independently of the other.

What is the probability that both children are girls, given that **at least one** of them is a girl?

# Two-Children Probability Problem

Consider a family with two children.  
Assume each child is equally likely to be a boy (B) or a girl (G), independently.

---

## Sample Space

The possible outcomes are:
$$
\Omega = \{BB, BG, GB, GG\}
$$

Each outcome has probability:
$$
P = \frac{1}{4}
$$

---

## Given Condition: At Least One Child is a Girl

The only outcome that does **not** satisfy this condition is $BB$.

So the reduced sample space becomes:
$$
\{BG, GB, GG\}
$$

These three outcomes are still equally likely.

---

## Required Probability

We want:
$$
P(\text{both girls} \mid \text{at least one girl})
$$

Only one outcome ($GG$) satisfies this.

Thus,
$$
P = \frac{1}{3}
$$

\[
\boxed{\frac{1}{3}}
\]

---

## Why This Can Be Confusing

A common mistake is to think the answer is $\frac{1}{2}$.  
That would be correct **only under a different condition**.

---

## Alternative Scenario

If instead we are told:
> "The older child is a girl"

Then the possible outcomes reduce to:
$$
\{GG, GB\}
$$

Now,
$$
P(\text{both girls}) = \frac{1}{2}
$$

---

## Key Insight

The result depends strongly on **how the information is given**:

- "At least one is a girl" → 3 possible cases → probability = $\frac{1}{3}$
- "Older child is a girl" → 2 possible cases → probability = $\frac{1}{2}$

---

## Conclusion

This example highlights an important idea in probability:

> Conditional probability depends not just on the event, but on the **way the condition is defined**.

# Q9) Information Available in a Discrete Random Variable

---

## 1. Probability Space

The sample space is
$$
\Omega = \{G_1, G_2, D_1, D_2\},
$$

with sigma-algebra
$$
\mathfrak F = \mathfrak P(\Omega),
$$

and probability measure:
$$
P(G_1)=0.50,\quad P(G_2)=0.20,\quad P(D_1)=0.10,\quad P(D_2)=0.20.
$$

Since
$$
0.50 + 0.20 + 0.10 + 0.20 = 1,
$$
this defines a valid probability space.

---

## 2. Sigma-algebra Generated by $X$

The random variable $X$ groups outcomes as:

- Good: $\{G_1, G_2\}$
- Defective: $\{D_1, D_2\}$

Thus,
$$
\sigma(X) = \{\varnothing, \Omega, \{G_1,G_2\}, \{D_1,D_2\}\}.
$$

---

## 3. Sigma-algebra Generated by $Y$

Since $Y$ assigns a unique value to each outcome, it distinguishes all elements.

Thus,
$$
\sigma(Y) = \mathfrak P(\Omega).
$$

---

## 4. Relationship Between $\sigma(X)$ and $\sigma(Y)$

We have
$$
\sigma(X) \subseteq \sigma(Y).
$$

This means $Y$ contains strictly more information than $X$.

- $X$: only quality (good/defective)
- $Y$: quality + machine identity

---

## 5. Marginal Distribution of $X$

$$
P(X=0) = 0.50 + 0.20 = 0.70,
$$

$$
P(X=1) = 0.10 + 0.20 = 0.30.
$$

---

## 6. Marginal Distribution of $Y$

$$
P(Y=1)=0.50,\quad P(Y=2)=0.20,\quad P(Y=3)=0.10,\quad P(Y=4)=0.20.
$$

---

## 7. Joint Distribution

Only consistent pairs occur:

$$
P(X=0,Y=1)=0.50,\quad P(X=0,Y=2)=0.20,
$$

$$
P(X=1,Y=3)=0.10,\quad P(X=1,Y=4)=0.20.
$$

All other joint probabilities are zero.

---

## 8. Conditional Probabilities

From $Y$ to $X$:
$$
P(X=1 \mid Y=3) = 1,\quad P(X=1 \mid Y=4) = 1.
$$

From $X$ to $Y$:
$$
P(Y=3 \mid X=1) = \frac{0.10}{0.30} = \frac{1}{3},
$$

$$
P(Y=4 \mid X=1) = \frac{0.20}{0.30} = \frac{2}{3}.
$$

---

## 9. Interpretation of Conditioning

- Conditioning on $Y$ gives **complete information** about $X$
- Conditioning on $X$ leaves **uncertainty** about $Y$

So $Y$ is more informative.

---

## 10. Information Content

- $\sigma(X)$ tells:
  - whether the item is good or defective

- $\sigma(Y)$ tells:
  - quality
  - machine used

Thus, $\sigma(Y)$ is strictly richer.

---

# Q10) Information in Continuous Random Variables

---

## 1. Probability Space

$$
\Omega = \mathbb{R}^2, \qquad \mathfrak{F} = \mathfrak{B}(\mathbb{R}^2)
$$

Each outcome is
$$
\omega = (t,p)
$$

The probability measure is defined via the joint density:
$$
f_{T,P}(t,p) = \frac{1}{4\pi} \exp\left(-\frac{t^2}{2} - \frac{p^2}{8}\right)
$$

Thus for any Borel set $A$:
$$
P(A) = \iint_A f_{T,P}(t,p)\, dt\, dp
$$

---

## 2. Sigma-algebra Generated by $X$

$$
X(t,p) = t
$$

Then
$$
X^{-1}(B) = B \times \mathbb{R}
$$

So
$$
\sigma(X) = \{ B \times \mathbb{R} : B \in \mathfrak{B}(\mathbb{R}) \}
$$

These are **vertical strips** in $\mathbb{R}^2$.

---

## 3. Sigma-algebra Generated by $Y$

$$
Y(t,p) = (t,p)
$$

Since $Y$ is the identity map:
$$
\sigma(Y) = \mathfrak{B}(\mathbb{R}^2)
$$

---

## 4. Information Comparison

$$
\sigma(X) \subset \sigma(Y)
$$

Strict inclusion holds because sets like
$$
[-1,1]\times[-2,2]
$$
depend on both coordinates.

**Interpretation:**
- $X$ → temperature only
- $Y$ → temperature + pressure

---

## 5. Marginal Density of $X$

Integrating out $p$:
$$
f_X(x) = \int_{-\infty}^{\infty} f_{T,P}(x,p)\, dp
$$

Gives:
$$
f_X(x) = \frac{1}{\sqrt{2\pi}} e^{-x^2/2}
$$

So
$$
X \sim N(0,1)
$$

---

## 6. Marginal Density of $Y$

$$
f_Y(t,p) = f_{T,P}(t,p)
$$

---

## 7. Probabilities

$$
P(X \le 0) = \frac{1}{2}
$$

$$
P(X > 1) = 1 - \Phi(1) \approx 0.1587
$$

$$
P(Y \in (-\infty,0]\times\mathbb{R}) = \frac{1}{2}
$$

Using independence:
$$
P(Y \in [-1,1]\times[-2,2]) = (2\Phi(1)-1)^2 \approx 0.466
$$

---

## 8. Conditional Distribution of $X \mid Y$

Since $X = t$ is determined by $Y=(t,p)$:
$$
X \mid Y=(t,p) \sim \delta_t
$$

(point mass at $t$)

---

## 9. Conditional Distribution of $Y \mid X$

Given $X=t$:
$$
Y \mid X=t \sim (t, P), \quad P \sim N(0,4)
$$

So:
- temperature fixed
- pressure still random

---

## 10. Interpretation

- Conditioning on $X$: know temperature only  
- Conditioning on $Y$: know both temperature and pressure  

Thus $Y$ provides strictly more information.

---

## 11. Functional Relationship

Define:
$$
\pi_1(t,p) = t
$$

Then:
$$
X = \pi_1 \circ Y
$$

So $X$ is a measurable function of $Y$, implying:
$$
\sigma(X) \subseteq \sigma(Y)
$$

---

## 12. Visualization Insight

- $\sigma(X)$ → vertical strips $B \times \mathbb{R}$
- $\sigma(Y)$ → all 2D regions

So:
- $X$ sees **1D structure**
- $Y$ sees **full 2D geometry**

---

## 13. Differential Entropy of $X$

For $X \sim N(0,1)$:
$$
h(X) = \frac{1}{2} \ln(2\pi e) \approx 1.4189
$$

---

## 14. Differential Entropy of $Y$

Since independent:
$$
h(Y) = h(T) + h(P)
$$

$$
h(Y) = \frac{1}{2}\ln(2\pi e) + \frac{1}{2}\ln(8\pi e)
$$

$$
h(Y) = \ln(2\pi e) + \ln 2
$$

$$
h(Y) \approx 3.5310
$$

---

## 15. Conditional Entropy $h(Y|X)$

Since only $P$ remains uncertain:
$$
h(Y|X) = h(P) = \frac{1}{2}\ln(8\pi e)
$$

$$
h(Y|X) \approx 2.1121
$$

**Interpretation:**
Knowing temperature leaves uncertainty only in pressure.

---

## 16. Behavior of $h(X|Y)$

Although $X$ is fully determined by $Y$, we do **not** get:
$$
h(X|Y) = 0
$$

Instead:
$$
h(X|Y) = -\infty
$$

---

## Key Insight

- Discrete case:
  $$
  H(X|Y) = 0 \quad \text{if } X \text{ is determined by } Y
  $$

- Continuous case:
  Deterministic dependence leads to **singular distributions**, causing entropy to diverge.

---

## Final Conclusion

- $X$ is a **coarse observation**
- $Y$ is a **refined observation**

$$
\sigma(X) \subseteq \sigma(Y), \quad h(X) < h(Y)
$$

- Knowing $X$ leaves uncertainty in $Y$
- Knowing $Y$ completely determines $X$

This example highlights the deeper structure of information in continuous probability spaces.
