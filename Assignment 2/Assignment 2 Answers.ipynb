# Emergency Transformer Failure Model

Let $X$ denote the number of emergency transformer failures reported in a single day.

The possible values are:
$$
X \in \{0,1,2,3\}
$$

with probabilities
$$
P(X=0)=0.50,
$$

$$
P(X=1)=0.30,
$$

$$
P(X=2)=0.15,
$$

$$
P(X=3)=0.05.
$$

---

## 1. Constructing the Probability Space

Take the sample space as
$$
\Omega = \{0,1,2,3\}.
$$

Let
$$
\mathfrak F = \mathfrak P(\Omega)
$$
be the power set of $\Omega$.

Define the probability measure $P$ by:
$$
P(\{0\})=0.50,
$$

$$
P(\{1\})=0.30,
$$

$$
P(\{2\})=0.15,
$$

$$
P(\{3\})=0.05.
$$

Since
$$
0.50+0.30+0.15+0.05=1,
$$
this defines a valid probability space.

---

## 2. Defining the Random Variable

Define the random variable
$$
X:\Omega \to \mathbb R
$$

by
$$
X(\omega)=\omega.
$$

Thus $X$ directly records the number of transformer failures occurring in a day.

---

## 3. Probability Mass Function (PMF)

The probability mass function is

$$
p_X(x)=P(X=x),
$$

given by

$$
p_X(x)=
\begin{cases}
0.50, & x=0, \\
0.30, & x=1, \\
0.15, & x=2, \\
0.05, & x=3, \\
0, & \text{otherwise}.
\end{cases}
$$

---

## 4. Expectation of $X$

The expectation is

$$
E[X]=\sum_x xP(X=x).
$$

Therefore,

$$
E[X]
=0(0.50)+1(0.30)+2(0.15)+3(0.05).
$$

Hence,

$$
E[X]
=0+0.30+0.30+0.15
=0.75.
$$

Thus,

$$
\boxed{E[X]=0.75}
$$

---

## 5. Variance of $X$

First compute

$$
E[X^2]
=\sum_x x^2P(X=x).
$$

So,

$$
E[X^2]
=0^2(0.50)+1^2(0.30)+2^2(0.15)+3^2(0.05).
$$

$$
=0+0.30+0.60+0.45
=1.35.
$$

Now,

$$
\mathrm{Var}(X)=E[X^2]-(E[X])^2.
$$

Thus,

$$
\mathrm{Var}(X)
=1.35-(0.75)^2.
$$

$$
=1.35-0.5625
=0.7875.
$$

Therefore,

$$
\boxed{\mathrm{Var}(X)=0.7875}
$$

---

## 6. Computing Probabilities

### Probability of at least one failure

$$
P(X\ge1)=P(X=1)+P(X=2)+P(X=3).
$$

$$
=0.30+0.15+0.05
=0.50.
$$

Hence,

$$
\boxed{P(X\ge1)=0.50}
$$

---

### Probability of at least two failures

$$
P(X\ge2)=P(X=2)+P(X=3).
$$

$$
=0.15+0.05
=0.20.
$$

Therefore,

$$
\boxed{P(X\ge2)=0.20}
$$

---

## 7. Practical Interpretation

- The expectation
$$
E[X]=0.75
$$
means that, on average, the district experiences approximately $0.75$ emergency transformer failures per day.

- The variance
$$
\mathrm{Var}(X)=0.7875
$$
measures the variability in the number of failures from day to day.

A larger variance would indicate that the daily number of failures fluctuates more significantly around the average value.

# Defective Item Proportion Model

Let $X$ denote the proportion of defective items in a large production batch.

Since $X$ represents a proportion,
$$
0 \le X \le 1.
$$

Assume that $X$ follows a Beta distribution with parameters
$$
\alpha = 2, \qquad \beta = 18.
$$

The density is
$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x(1-x)^{17}, & 0<x<1, \\
0, & \text{otherwise}.
\end{cases}
$$

---

## 1. Constructing the Probability Space

Take
$$
\Omega = [0,1],
$$

with sigma-algebra
$$
\mathfrak F = \mathfrak B([0,1]),
$$

the Borel sigma-algebra on $[0,1]$.

Define the probability measure $P$ by
$$
P(A)=\int_A f(x)\,dx,
\qquad A\in\mathfrak F.
$$

---

## 2. Defining the Random Variable

Define
$$
X:\Omega \to \mathbb R
$$

by
$$
X(\omega)=\omega.
$$

Thus $X$ records the defective proportion in the batch.

---

## 3. Probability Density Function

The probability density function is

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x(1-x)^{17}, & 0<x<1, \\
0, & \text{otherwise}.
\end{cases}
$$

For Beta distributions,
$$
B(\alpha,\beta)
=
\frac{\Gamma(\alpha)\Gamma(\beta)}
{\Gamma(\alpha+\beta)}.
$$

Hence,
$$
B(2,18)
=
\frac{\Gamma(2)\Gamma(18)}
{\Gamma(20)}.
$$

---

## 4. Verification that $f$ is a Valid Density

A valid density must satisfy:

### (i) Nonnegativity

For $0<x<1$,
$$
x>0
\quad\text{and}\quad
(1-x)^{17}>0,
$$

so
$$
f(x)\ge0.
$$

---

### (ii) Total Probability Equals 1

We compute
$$
\int_{-\infty}^{\infty} f(x)\,dx.
$$

Since $f(x)=0$ outside $(0,1)$,
$$
\int_0^1 \frac{1}{B(2,18)}x(1-x)^{17}\,dx.
$$

Using the definition of the Beta function,
$$
\int_0^1 x^{2-1}(1-x)^{18-1}\,dx
=
B(2,18).
$$

Therefore,
$$
\int_0^1 f(x)\,dx
=
\frac{1}{B(2,18)}B(2,18)
=1.
$$

Hence $f$ is a valid probability density function.

---

## 5. Expectation of $X$

For a Beta$(\alpha,\beta)$ distribution,
$$
E[X]
=
\frac{\alpha}{\alpha+\beta}.
$$

Thus,
$$
E[X]
=
\frac{2}{2+18}
=
\frac{2}{20}
=
0.10.
$$

Therefore,
$$
\boxed{E[X]=0.10}
$$

---

## 6. Variance of $X$

For a Beta$(\alpha,\beta)$ distribution,
$$
\mathrm{Var}(X)
=
\frac{\alpha\beta}
{(\alpha+\beta)^2(\alpha+\beta+1)}.
$$

Hence,
$$
\mathrm{Var}(X)
=
\frac{2(18)}
{20^2(21)}.
$$

$$
=
\frac{36}{8400}
=
\frac{3}{700}
\approx 0.00429.
$$

Thus,
$$
\boxed{\mathrm{Var}(X)=\frac{3}{700}\approx0.00429}
$$

---

## 7. Probability that $X \le 0.10$

We compute
$$
P(X\le0.10)
=
\int_0^{0.10} f(x)\,dx.
$$

Thus,
$$
P(X\le0.10)
=
\frac{1}{B(2,18)}
\int_0^{0.10} x(1-x)^{17}\,dx.
$$

Therefore,
$$
\boxed{
P(X\le0.10)
=
\frac{1}{B(2,18)}
\int_0^{0.10} x(1-x)^{17}\,dx
}
$$

---

## 8. Probability that $0.05 \le X \le 0.15$

Similarly,
$$
P(0.05\le X\le0.15)
=
\int_{0.05}^{0.15} f(x)\,dx.
$$

Hence,
$$
P(0.05\le X\le0.15)
=
\frac{1}{B(2,18)}
\int_{0.05}^{0.15} x(1-x)^{17}\,dx.
$$

Therefore,
$$
\boxed{
P(0.05\le X\le0.15)
=
\frac{1}{B(2,18)}
\int_{0.05}^{0.15} x(1-x)^{17}\,dx
}
$$

---

## 9. Why the Beta Distribution is Appropriate

The Beta distribution is suitable because:

- it is defined on the interval $[0,1]$,
- it naturally models proportions and probabilities,
- its shape can vary widely depending on the parameters,
- it can represent skewed production quality behavior.

With $\alpha=2$ and $\beta=18$, the distribution is concentrated near small values, reflecting that defective proportions are usually low in a well-controlled manufacturing process.

---

## 10. Interpretation of Expectation and Variance

The expectation
$$
E[X]=0.10
$$

means that the average defective proportion is about $10\%$.

The variance
$$
\mathrm{Var}(X)\approx0.00429
$$

measures how much the defective proportion fluctuates from day to day.

A small variance indicates that the manufacturing quality is relatively stable and consistent across production runs.

# Triangular Distribution Model for Generator Installation Time

Let $X$ denote the number of hours required to install the backup generator.

Assume that $X$ follows a Triangular distribution with parameters

$$
a=4, \qquad b=10, \qquad c=6,
$$

where

- $a$ = minimum possible time,
- $b$ = maximum possible time,
- $c$ = most likely completion time.

---

## 1. Support of $X$

The support of a Triangular distribution is the interval

$$
[a,b].
$$

Hence,

$$
\boxed{4 \le X \le 10}
$$

---

## 2. Probability Density Function

For a Triangular$(a,b,c)$ distribution, the density is

$$
f_X(x)=
\begin{cases}
\dfrac{2(x-a)}{(b-a)(c-a)}, & a\le x\le c, \\[10pt]
\dfrac{2(b-x)}{(b-a)(b-c)}, & c < x\le b, \\[10pt]
0, & \text{otherwise}.
\end{cases}
$$

Substituting

$$
a=4,\qquad b=10,\qquad c=6,
$$

gives

$$
f_X(x)=
\begin{cases}
\dfrac{x-4}{6}, & 4\le x\le 6, \\[10pt]
\dfrac{10-x}{12}, & 6 < x\le 10, \\[10pt]
0, & \text{otherwise}.
\end{cases}
$$

---

## 3. Verification that $f_X$ Integrates to $1$

We compute

$$
\int_{-\infty}^{\infty} f_X(x)\,dx.
$$

Since the density is zero outside $[4,10]$,

$$
\int_4^6 \frac{x-4}{6}\,dx
+
\int_6^{10} \frac{10-x}{12}\,dx.
$$

### First Integral

$$
\int_4^6 \frac{x-4}{6}\,dx
=
\frac16\int_4^6 (x-4)\,dx.
$$

$$
=
\frac16\left[\frac{(x-4)^2}{2}\right]_4^6
=
\frac16\cdot\frac{4}{2}
=
\frac13.
$$

### Second Integral

$$
\int_6^{10} \frac{10-x}{12}\,dx
=
\frac1{12}\int_6^{10}(10-x)\,dx.
$$

$$
=
\frac1{12}\left[10x-\frac{x^2}{2}\right]_6^{10}
=
\frac23.
$$

Therefore,

$$
\frac13+\frac23=1.
$$

Hence,

$$
\boxed{\int_{-\infty}^{\infty} f_X(x)\,dx=1}
$$

so $f_X$ is a valid probability density function.

---

## 4. Expectation $E[X]$

For a Triangular distribution,

$$
E[X]=\frac{a+b+c}{3}.
$$

Thus,

$$
E[X]
=
\frac{4+10+6}{3}
=
\frac{20}{3}.
$$

Hence,

$$
\boxed{E[X]=\frac{20}{3}\approx 6.67 \text{ hours}}
$$

---

## 5. Variance $\mathrm{Var}(X)$

For a Triangular distribution,

$$
\mathrm{Var}(X)
=
\frac{
a^2+b^2+c^2-ab-ac-bc
}{18}.
$$

Substituting the values:

$$
\mathrm{Var}(X)
=
\frac{
4^2+10^2+6^2-(4)(10)-(4)(6)-(10)(6)
}{18}.
$$

$$
=
\frac{
16+100+36-40-24-60
}{18}.
$$

$$
=
\frac{28}{18}
=
\frac{14}{9}.
$$

Therefore,

$$
\boxed{\mathrm{Var}(X)=\frac{14}{9}\approx1.56}
$$

---

## 6. Interpretation of $a$, $b$, and $c$

In this context:

- $a=4$ hours represents the shortest possible installation time,
- $b=10$ hours represents the longest possible installation time,
- $c=6$ hours represents the most likely completion time according to expert judgment.

---

## 7. Why the Triangular Distribution is Reasonable

The Triangular distribution is useful when:

- exact historical data are limited,
- experts can estimate minimum, maximum, and most likely values,
- the process has a natural lower and upper bound.

In project management and engineering estimation, this distribution provides a simple way to model uncertainty using only a few intuitive parameters.

---

## 8. Probability that Installation Takes at Most $5$ Hours

We compute

$$
P(X\le5)
=
\int_4^5 \frac{x-4}{6}\,dx.
$$

Thus,

$$
P(X\le5)
=
\frac16\left[\frac{(x-4)^2}{2}\right]_4^5.
$$

$$
=
\frac16\cdot\frac12
=
\frac1{12}.
$$

Hence,

$$
\boxed{P(X\le5)=\frac1{12}\approx0.0833}
$$

---

## 9. Probability that Installation Takes Between $5$ and $8$ Hours

We compute

$$
P(5\le X\le 8)
=
\int_5^6 \frac{x-4}{6}\,dx
+
\int_6^8 \frac{10-x}{12}\,dx.
$$

### First Part

$$
\int_5^6 \frac{x-4}{6}\,dx
=
\frac16\left[\frac{(x-4)^2}{2}\right]_5^6.
$$

$$
=
\frac16\left(2-\frac12\right)
=
\frac14.
$$

### Second Part

$$
\int_6^8 \frac{10-x}{12}\,dx
=
\frac1{12}\left[10x-\frac{x^2}{2}\right]_6^8.
$$

$$
=
\frac1{12}(48-42)
=
\frac12.
$$

Therefore,

$$
P(5\le X\le8)
=
\frac14+\frac12
=
\frac34.
$$

Hence,

$$
\boxed{P(5\le X\le8)=\frac34=0.75}
$$

---

## 10. Use in Project Planning and Sensitivity Analysis

This model helps project planners estimate likely completion times and evaluate scheduling risk.

For example:

  $$
  P(X\le5)
  $$
  help estimate the likelihood of early completion,

Such models are useful in resource allocation, budgeting, contingency planning, and sensitivity analysis for construction projects.

# PERT Distribution Model for Software Integration Time

Let $X$ denote the number of days required to complete the software integration task.

The project manager estimates:

$$
a=8, \qquad b=20, \qquad c=11,
$$

where

- $a$ = shortest possible completion time,
- $b$ = longest possible completion time,
- $c$ = most likely completion time.

The PERT distribution is a scaled Beta distribution on the interval $[a,b]$.

---

## 1. Compute the Shape Parameters $\alpha$ and $\beta$

The parameters are defined by

$$
\alpha
=
1+4\left(\frac{c-a}{b-a}\right),
$$

and

$$
\beta
=
1+4\left(\frac{b-c}{b-a}\right).
$$

Substituting the values:

$$
\alpha
=
1+4\left(\frac{11-8}{20-8}\right)
=
1+4\left(\frac{3}{12}\right).
$$

$$
=
1+1
=
2.
$$

Similarly,

$$
\beta
=
1+4\left(\frac{20-11}{20-8}\right)
=
1+4\left(\frac{9}{12}\right).
$$

$$
=
1+3
=
4.
$$

Therefore,

$$
\boxed{\alpha=2,\qquad \beta=4}
$$

---

## 2. Support of $X$

Since the task duration cannot be less than $8$ days or exceed $20$ days,

$$
\boxed{8\le X\le 20}
$$

Thus the support is the interval

$$
[8,20].
$$

---

## 3. Probability Density Function

The PERT density is

$$
f(x)=
\frac{
(x-a)^{\alpha-1}(b-x)^{\beta-1}
}{
B(\alpha,\beta)(b-a)^{\alpha+\beta-1}
},
\qquad a\le x\le b.
$$

Substituting

$$
a=8,\qquad b=20,\qquad \alpha=2,\qquad \beta=4,
$$

gives

$$
f_X(x)
=
\frac{
(x-8)^{1}(20-x)^{3}
}{
B(2,4)(12)^5
},
\qquad 8\le x\le20.
$$

and

$$
f_X(x)=0
\qquad \text{otherwise}.
$$

Hence,

$$
\boxed{
f_X(x)
=
\frac{
(x-8)(20-x)^3
}{
B(2,4)\,12^5
},
\qquad 8\le x\le20
}
$$

---

## 4. Probability Space

A suitable probability space is

$$
\Omega=[8,20],
$$

with sigma-algebra

$$
\mathfrak F=\mathfrak B([8,20]),
$$

the Borel sigma-algebra on $[8,20]$.

The probability measure is defined by

$$
P(A)=\int_A f_X(x)\,dx,
\qquad A\in\mathfrak F.
$$

---

## 5. Definition of the Random Variable

Define

$$
X:\Omega\to\mathbb R
$$

by

$$
X(\omega)=\omega.
$$

Thus $X$ records the project completion time in days.

---

## 6. Expectation of $X$

For the PERT model,

$$
E[X]=\frac{a+4c+b}{6}.
$$

Substituting the values:

$$
E[X]
=
\frac{8+4(11)+20}{6}.
$$

$$
=
\frac{8+44+20}{6}
=
\frac{72}{6}
=
12.
$$

Therefore,

$$
\boxed{E[X]=12 \text{ days}}
$$

The PERT expectation places more weight on the most likely value $c$.

---

## 7. Variance of $X$

Using the approximation

$$
\mathrm{Var}(X)
=
\frac{(E[X]-a)(b-E[X])}{7},
$$

and using

$$
E[X]=12,
$$

we obtain

$$
\mathrm{Var}(X)
=
\frac{(12-8)(20-12)}{7}.
$$

$$
=
\frac{(4)(8)}{7}
=
\frac{32}{7}.
$$

Hence,

$$
\boxed{
\mathrm{Var}(X)=\frac{32}{7}\approx4.57
}
$$

---

## 8. Standard Deviation

The standard deviation is

$$
\sigma=\sqrt{\mathrm{Var}(X)}.
$$

Thus,

$$
\sigma
=
\sqrt{\frac{32}{7}}
\approx2.14.
$$

Therefore,

$$
\boxed{\sigma\approx2.14 \text{ days}}
$$

---

## 9. Why PERT May Be Better than a Triangular Distribution

The Triangular distribution uses straight-line densities and can produce sharp corners at the mode.

The PERT distribution smooths the density using a Beta distribution, giving:

- a more realistic shape,
- smoother probability transitions,
- less sensitivity to extreme values,
- better behavior for project scheduling applications.

Also, the PERT mean

$$
E[X]=\frac{a+4c+b}{6}
$$

gives extra weight to the most likely value $c$, which often matches expert judgment more naturally than the simple average used in some simpler models.

Thus PERT models are commonly preferred in project management and risk analysis.

---

## 10. Interpretation of the Parameters

### Minimum Time $a=8$

This is the shortest possible completion time assuming everything proceeds ideally.

---

### Maximum Time $b=20$

This represents the worst-case completion time under major delays or complications.

---

### Most Likely Time $c=11$

This is the duration considered most realistic by experts.

---

### Expected Time $E[X]=12$

This is the average estimated completion time after incorporating uncertainty in the model.

### Standard Deviation $\sigma\approx2.14$

# Q7) If the PERT Expectation is to Match Our Intuition

A project manager models the duration $X$ of a task by a PERT distribution with parameters $a$, $b$, and $c$, where $a$ is the minimum time, $b$ is the maximum time, and $c$ is the most likely time.

Suppose the minimum time is $a=5$ days and the most likely time is $c=11$ days. If the expected time computed from the PERT model is also to be the same as the most likely time:

1. Determine the value of the maximum time $b$.

2. What happens to the variance?

# PERT Distribution Special Case

Suppose a task duration $X$ follows a PERT distribution with parameters

$$
a=5, \qquad c=11,
$$

where

- $a$ = minimum completion time,
- $c$ = most likely completion time,
- $b$ = maximum completion time.

The PERT expected value is

$$
E[X]=\frac{a+4c+b}{6}.
$$

We are told that the expected time is equal to the most likely time:

$$
E[X]=c.
$$

---

## 1. Determine the Value of $b$

Since

$$
E[X]=c,
$$

we substitute into the PERT expectation formula:

$$
\frac{a+4c+b}{6}=c.
$$

Substituting

$$
a=5,\qquad c=11,
$$

gives

$$
\frac{5+4(11)+b}{6}=11.
$$

Simplifying,

$$
\frac{5+44+b}{6}=11.
$$

$$
\frac{49+b}{6}=11.
$$

Multiply both sides by $6$:

$$
49+b=66.
$$

Therefore,

$$
b=66-49.
$$

Hence,

$$
\boxed{b=17}
$$

---

## 2. What Happens to the Variance?

Using the common PERT approximation,

$$
\mathrm{Var}(X)
=
\frac{(E[X]-a)(b-E[X])}{7}.
$$

Since

$$
E[X]=11,
$$

and

$$
a=5,\qquad b=17,
$$

we obtain

$$
\mathrm{Var}(X)
=
\frac{(11-5)(17-11)}{7}.
$$

$$
=
\frac{(6)(6)}{7}
=
\frac{36}{7}.
$$

Therefore,

$$
\boxed{
\mathrm{Var}(X)=\frac{36}{7}\approx5.14
}
$$

and the standard deviation is

$$
\sigma
=
\sqrt{\frac{36}{7}}
=
\frac{6}{\sqrt7}
\approx2.27.
$$

---

## Interpretation

In this case,

$$
E[X]=c=11,
$$

which means the expected completion time coincides exactly with the most likely completion time.

This happens because the interval is symmetric around $11$:

$$
11-5=17-11=6.
$$

Thus the PERT distribution becomes balanced around the mode, producing a distribution with no directional skewness.

The variance remains positive because there is still uncertainty in completion time between the minimum and maximum values.
