## Conditional Expectation and Information in Fraud Detection

Let

$$
\Omega=\{\omega_1,\omega_2,\omega_3,\omega_4\},
$$

where

- $\omega_1$: low-risk customer, legitimate transaction,
- $\omega_2$: low-risk customer, fraudulent transaction,
- $\omega_3$: high-risk customer, legitimate transaction,
- $\omega_4$: high-risk customer, fraudulent transaction.

The probabilities are

$$
P(\{\omega_1\})=0.50,
$$

$$
P(\{\omega_2\})=0.10,
$$

$$
P(\{\omega_3\})=0.20,
$$

$$
P(\{\omega_4\})=0.20.
$$

---

## 1. Probability Space

The probability space is

$$
(\Omega,\mathfrak F,P),
$$

where

$$
\Omega=\{\omega_1,\omega_2,\omega_3,\omega_4\},
$$

$$
\mathfrak F=\mathfrak P(\Omega),
$$

and $P$ is defined by the probabilities above.

Since

$$
0.50+0.10+0.20+0.20=1,
$$

this is a valid probability measure.

---

## 2. Sigma-Algebra Generated by $Y$

The random variable $Y$ is

$$
Y(\omega_1)=0,
\qquad
Y(\omega_2)=0,
$$

$$
Y(\omega_3)=1,
\qquad
Y(\omega_4)=1.
$$

Thus the partition induced by $Y$ is

$$
\{\omega_1,\omega_2\},
\qquad
\{\omega_3,\omega_4\}.
$$

Hence,

$$
\sigma(Y)
=
\{
\varnothing,
\Omega,
\{\omega_1,\omega_2\},
\{\omega_3,\omega_4\}
\}.
$$

Therefore,

$$
\boxed{
\sigma(Y)=
\{
\varnothing,
\Omega,
\{\omega_1,\omega_2\},
\{\omega_3,\omega_4\}
\}
}
$$

---

## 3. Why $E[X\mid Y]$ Must be Constant on Each Block

By definition,

$$
E[X\mid Y]
=
E[X\mid \sigma(Y)]
$$

must be $\sigma(Y)$-measurable.

A $\sigma(Y)$-measurable function can only depend on whether

$$
Y=0
\quad\text{or}\quad
Y=1.
$$

Therefore it must be constant on each atom of $\sigma(Y)$:

$$
\{\omega_1,\omega_2\}
$$

and

$$
\{\omega_3,\omega_4\}.
$$

So there exist constants $a$ and $b$ such that

$$
E[X\mid Y]
=
a\,\mathbf 1_{\{Y=0\}}
+
b\,\mathbf 1_{\{Y=1\}}.
$$

---

## 4. Compute $E[X\mid Y]$

### Case 1: $Y=0$

This corresponds to the set

$$
\{\omega_1,\omega_2\}.
$$

We compute

$$
P(Y=0)=0.50+0.10=0.60.
$$

Since fraud corresponds to $X=1$,

$$
E[X\mid Y=0]
=
P(X=1\mid Y=0).
$$

Thus

$$
E[X\mid Y=0]
=
\frac{0.10}{0.60}
=
\frac16.
$$

---

### Case 2: $Y=1$

This corresponds to

$$
\{\omega_3,\omega_4\}.
$$

We compute

$$
P(Y=1)=0.20+0.20=0.40.
$$

Thus

$$
E[X\mid Y=1]
=
P(X=1\mid Y=1)
=
\frac{0.20}{0.40}
=
\frac12.
$$

Therefore,

$$
\boxed{
E[X\mid Y]
=
\frac16\mathbf1_{\{Y=0\}}
+
\frac12\mathbf1_{\{Y=1\}}
}
$$

---

## 5. Verification of the Defining Property

We verify

$$
\int_A E[X\mid Y]\,dP
=
\int_A X\,dP
$$

for every

$$
A\in\sigma(Y).
$$

It suffices to check the nontrivial atoms.

---

### Set $A=\{\omega_1,\omega_2\}$

Left side:

$$
\int_A E[X\mid Y]\,dP
=
\frac16(0.60)
=
0.10.
$$

Right side:

$$
\int_A X\,dP
=
0(0.50)+1(0.10)
=
0.10.
$$

Hence both sides agree.

---

### Set $A=\{\omega_3,\omega_4\}$

Left side:

$$
\int_A E[X\mid Y]\,dP
=
\frac12(0.40)
=
0.20.
$$

Right side:

$$
\int_A X\,dP
=
0(0.20)+1(0.20)
=
0.20.
$$

Again both sides agree.

Thus the defining property holds.

---

## 6. Conditional PMF $p_{X|Y}(x\mid y)$

### For $Y=0$

$$
P(X=0\mid Y=0)
=
\frac{0.50}{0.60}
=
\frac56,
$$

$$
P(X=1\mid Y=0)
=
\frac{0.10}{0.60}
=
\frac16.
$$

---

### For $Y=1$

$$
P(X=0\mid Y=1)
=
\frac{0.20}{0.40}
=
\frac12,
$$

$$
P(X=1\mid Y=1)
=
\frac{0.20}{0.40}
=
\frac12.
$$

Hence

$$
\boxed{
p_{X|Y}(x\mid0)=
\begin{cases}
\frac56,&x=0,\\
\frac16,&x=1,
\end{cases}
}
$$

and

$$
\boxed{
p_{X|Y}(x\mid1)=
\begin{cases}
\frac12,&x=0,\\
\frac12,&x=1.
\end{cases}
}
$$

---

## 7. Verification of the Conditional Expectation Formula

We verify

$$
E[X\mid Y=y]
=
\sum_{x\in\{0,1\}}
x\,p_{X|Y}(x\mid y).
$$

---

### For $y=0$

$$
E[X\mid Y=0]
=
0\cdot\frac56
+
1\cdot\frac16
=
\frac16.
$$

---

### For $y=1$

$$
E[X\mid Y=1]
=
0\cdot\frac12
+
1\cdot\frac12
=
\frac12.
$$

Thus the identity holds.

## 8. Mean Square Error (MSE)

Using the conditional expectation predictor,

$$
\mathrm{MSE}
=
E[(X-E[X\mid Y])^2].
$$

Using the law

$$
E[(X-E[X\mid Y])^2]
=
E[\mathrm{Var}(X\mid Y)].
$$

---

### Conditional Variances

For $Y=0$:

$$
\mathrm{Var}(X\mid Y=0)
=
\frac16\left(1-\frac16\right)
=
\frac{5}{36}.
$$

For $Y=1$:

$$
\mathrm{Var}(X\mid Y=1)
=
\frac12\left(1-\frac12\right)
=
\frac14.
$$

Now average using probabilities:

$$
\mathrm{MSE}
=
0.60\cdot\frac{5}{36}
+
0.40\cdot\frac14.
$$

$$
=
\frac{1}{12}+\frac{1}{10}
=
\frac{11}{60}.
$$

Therefore,

$$
\boxed{
\mathrm{MSE}=\frac{11}{60}\approx0.1833
}
$$

---

## 9. MSE Using the Global Average Fraud Rate

First compute

$$
E[X]
=
0(0.70)+1(0.30)
=
0.30.
$$

Using the constant estimator $0.30$ for every transaction,

$$
\mathrm{MSE}
=
E[(X-0.30)^2].
$$

Since $X$ is Bernoulli with parameter $0.30$,

$$
\mathrm{MSE}
=
\mathrm{Var}(X)
=
0.30(0.70)
=
0.21.
$$

Hence,

$$
\boxed{
\mathrm{MSE}=0.21
}
$$

which is larger than

$$
\frac{11}{60}\approx0.1833.
$$

Thus conditioning on $Y$ improves prediction accuracy.

---

### Relation to the Law of Total Variance

The law of total variance states

$$
\mathrm{Var}(X)
=
E[\mathrm{Var}(X\mid Y)]
+
\mathrm{Var}(E[X\mid Y]).
$$

Here,

$$
0.21
=
0.1833
+
0.0267.
$$

Thus the reduction in MSE comes from the variability explained by the customer risk class.

## 10. Entropy and Mutual Information

We use base-2 logarithms.

---

### (a) Entropy $H(X)$

The marginal distribution of $X$ is

$$
P(X=0)=0.70,
\qquad
P(X=1)=0.30.
$$

Thus

$$
H(X)
=
-0.70\log_2(0.70)
-0.30\log_2(0.30).
$$

Numerically,

$$
\boxed{
H(X)\approx0.8813\text{ bits}
}
$$

This is the bank's initial uncertainty regarding fraud.

---

### (b) Conditional Entropy $H(X\mid Y)$

Using

$$
H(X\mid Y)
=
\sum_y P(Y=y)H(X\mid Y=y).
$$

---

### For $Y=0$

$$
H(X\mid Y=0)
=
-\frac56\log_2\frac56
-\frac16\log_2\frac16.
$$

Numerically,

$$
H(X\mid Y=0)\approx0.6500.
$$

---

### For $Y=1$

$$
H(X\mid Y=1)
=
- \frac12\log_2\frac12
- \frac12\log_2\frac12
=
1.
$$

Therefore,

$$
H(X\mid Y)
=
0.60(0.6500)+0.40(1).
$$

$$
\boxed{
H(X\mid Y)\approx0.7900\text{ bits}
}
$$

---

### (c) Mutual Information $I(X;Y)$

By definition,

$$
I(X;Y)
=
H(X)-H(X\mid Y).
$$

Thus

$$
I(X;Y)
=
0.8813-0.7900.
$$

Therefore,

$$
\boxed{
I(X;Y)\approx0.0913\text{ bits}
}
$$

---

### (d) Percentage of Uncertainty Removed

The percentage reduction is

$$
\frac{I(X;Y)}{H(X)}\times100\%.
$$

Thus

$$
\frac{0.0913}{0.8813}\times100\%
\approx10.36\%.
$$

Hence,

$$
\boxed{
\text{About }10.4\%\text{ of the uncertainty about fraud is removed by knowing the customer's risk class.}
}
$$

This means that customer risk classification provides useful predictive information, although substantial uncertainty still remains.

# Q2) Conditional Density of the Multivariate Gaussian

Let $\mathbf{Z}$ be the concatenated random matrix
$\mathbf{Z} = \begin{bmatrix} \mathbf{X} \\ \mathbf{Y} \end{bmatrix} \in \mathbb{R}^{n+m}$.


The random variable $\mathbf{Z}$ is said to have a **joint multivariate normal distribution** if every linear combination of its components follows a univariate normal distribution.

For $\mathbf{Z}$ to possess a **joint density function** (meaning its distribution is absolutely continuous with respect to the Lebesgue measure on $\mathbb{R}^{n+m}$), its covariance matrix must be non-singular (positive definite).


The joint density function $f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z})$ for $\mathbf{z} \in \mathbb{R}^{n+m}$ is given by:

$$f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z}) = \frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}|}} \exp\left( -\frac{1}{2} (\mathbf{z} - \boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1} (\mathbf{z} - \boldsymbol{\mu}) \right).$$

Here:

* **Mean ($\boldsymbol{\mu}$):** $\boldsymbol{\mu} = E[\mathbf{Z}] = \begin{bmatrix} \boldsymbol{\mu}_X \\ \boldsymbol{\mu}_Y \end{bmatrix} \in \mathbb{R}^{n+m}$, where $\boldsymbol{\mu}_X \in \mathbb{R}^n$ and $\boldsymbol{\mu}_Y \in \mathbb{R}^m$.
* **Covariance Matrix ($\boldsymbol{\Sigma}$):** A symmetric, positive definite matrix $\boldsymbol{\Sigma} \in \mathbb{R}^{(n+m) \times (n+m)}$ partitioned as:
$$\boldsymbol{\Sigma} = \begin{bmatrix} \boldsymbol{\Sigma}_{XX} & \boldsymbol{\Sigma}_{XY} \\ \boldsymbol{\Sigma}_{YX} & \boldsymbol{\Sigma}_{YY} \end{bmatrix}$$
    * $\boldsymbol{\Sigma}_{XX} = \text{Var}(\mathbf{X})$ (size $n \times n$)
    * $\boldsymbol{\Sigma}_{YY} = \text{Var}(\mathbf{Y})$ (size $m \times m$)
    * $\boldsymbol{\Sigma}_{XY} = \text{Cov}(\mathbf{X}, \mathbf{Y}) = \boldsymbol{\Sigma}_{YX}^T$ (size $n \times m$)
* **Determinant ($|\boldsymbol{\Sigma}|$):** The normalization factor ensuring the total integral is 1.

## 1. Joint Distribution

Since $X$ and $Q$ are independent Gaussian random variables and

$$
Y=X+Q,
$$

the vector $(X,Y)$ is jointly Gaussian.

We first compute the mean and variance of $Y$.

Since

$$
E[X]=0, \qquad E[Q]=0,
$$

we have

$$
E[Y]=E[X+Q]=0.
$$

Also,

$$
\mathrm{Var}(Y)=\mathrm{Var}(X+Q).
$$

Because $X$ and $Q$ are independent,

$$
\mathrm{Var}(Y)=\mathrm{Var}(X)+\mathrm{Var}(Q)=1+4=5.
$$

Hence

$$
Y\sim \mathscr{N}(0,5).
$$

Now compute the covariance:

$$
\mathrm{Cov}(X,Y)=\mathrm{Cov}(X,X+Q).
$$

Using linearity,

$$
\mathrm{Cov}(X,Y)
=
\mathrm{Var}(X)+\mathrm{Cov}(X,Q).
$$

Since $X$ and $Q$ are independent,

$$
\mathrm{Cov}(X,Q)=0,
$$

thus

$$
\boxed{\mathrm{Cov}(X,Y)=1.}
$$

Therefore the covariance matrix of $(X,Y)$ is

$$
\Sigma=
\begin{bmatrix}
1 & 1 \\
1 & 5
\end{bmatrix}.
$$

So

$$
(X,Y)\sim \mathscr{N}
\left(
\begin{bmatrix}
0\\
0
\end{bmatrix},
\begin{bmatrix}
1 & 1\\
1 & 5
\end{bmatrix}
\right).
$$

## 2. Information Geometry

The variable $X$ is **not** $\sigma(Y)$-measurable because knowing $Y$ does not uniquely determine $X$.

Indeed,

$$
Y=X+Q.
$$

Even if the engineer observes

$$
Y=5,
$$

there are infinitely many possibilities:

* the spindle temperature may be high while the cutting load is moderate,
* or the spindle temperature may be moderate while the cutting load is extremely high.

For example:

$$
5 = 5 + 0,
$$

or

$$
5 = 1 + 4,
$$

or

$$
5 = -2 + 7.
$$

Thus the observation $Y$ mixes together two independent effects:

* thermal behavior ($X$),
* mechanical load ($Q$).

Therefore the engineer cannot recover the true temperature exactly from the power measurement alone.

## 3. Conditional Expectation (Best Estimate)

For jointly Gaussian variables,

$$
E[X\mid Y=y]
=
\mu_X
+
\frac{\mathrm{Cov}(X,Y)}{\mathrm{Var}(Y)}
(y-\mu_Y).
$$

Since

$$
\mu_X=0,
\qquad
\mu_Y=0,
\qquad
\mathrm{Cov}(X,Y)=1,
\qquad
\mathrm{Var}(Y)=5,
$$

we obtain

$$
E[X\mid Y=y]
=
\frac{1}{5}y.
$$

Hence

$$
\boxed{
E[X\mid Y]
=
\frac15 Y.
}
$$

This can be written as

$$
E[X\mid Y]
=
\frac{\text{signal variance}}{\text{total variance}}
\,Y.
$$

Indeed,

* signal variance:
  $$
  \mathrm{Var}(X)=1,
  $$

* total observed variance:
  $$
  \mathrm{Var}(Y)=\mathrm{Var}(X)+\mathrm{Var}(Q)=5.
  $$

Thus

$$
E[X\mid Y]
=
\frac{1}{1+4}Y.
$$

So only one-fifth of the observed power fluctuation is attributed to temperature.

---

## 4. Verification of the Residual

Define

$$
\widehat X = E[X\mid Y]=\frac15Y.
$$

The residual error is

$$
Z=X-\widehat X.
$$

Substitute:

$$
Z=X-\frac15Y.
$$

Since

$$
Y=X+Q,
$$

we get

$$
Z
=
X-\frac15(X+Q)
=
\frac45X-\frac15Q.
$$

### Mean of the residual

$$
E[Z]
=
\frac45E[X]-\frac15E[Q].
$$

Since both means are zero,

$$
\boxed{E[Z]=0.}
$$

---

### Orthogonality Principle

Now compute

$$
\mathrm{Cov}(Z,Y).
$$

Using linearity,

$$
\mathrm{Cov}(Z,Y)
=
\mathrm{Cov}\left(X-\frac15Y,Y\right).
$$

Hence

$$
\mathrm{Cov}(Z,Y)
=
\mathrm{Cov}(X,Y)
-
\frac15\mathrm{Var}(Y).
$$

Substitute values:

$$
\mathrm{Cov}(Z,Y)
=
1-\frac15(5)=0.
$$

Therefore

$$
\boxed{\mathrm{Cov}(Z,Y)=0.}
$$

Thus the estimation error is uncorrelated with the observation.

This is the Orthogonality Principle.

## 5. Information Theory: Differential Entropy

### Entropy of $X$

For a Gaussian random variable,

$$
h(X)=\frac12\ln(2\pi e\sigma^2).
$$

Since $\sigma^2=1$,

$$
h(X)=\frac12\ln(2\pi e).
$$

Numerically,

$$
\boxed{
h(X)\approx 1.4189.
}
$$

---

### Conditional Entropy $h(X\mid Y)$

The conditional variance is

$$
\sigma_{X\mid Y}^2
=
\sigma_X^2
-
\frac{\mathrm{Cov}(X,Y)^2}{\mathrm{Var}(Y)}.
$$

Substitute values:

$$
\sigma_{X\mid Y}^2
=
1-\frac{1^2}{5}
=
\frac45.
$$

Therefore

$$
h(X\mid Y)
=
\frac12\ln\left(2\pi e\cdot\frac45\right).
$$

Numerically,

$$
\boxed{
h(X\mid Y)\approx 1.3073.
}
$$

---

## 6. Mutual Information

The mutual information is

$$
I(X;Y)=h(X)-h(X\mid Y).
$$

Thus

$$
I(X;Y)
=
\frac12\ln(2\pi e)
-
\frac12\ln\left(2\pi e\cdot\frac45\right).
$$

Simplify:

$$
I(X;Y)
=
\frac12\ln\left(\frac54\right).
$$

Hence

$$
\boxed{
I(X;Y)\approx 0.1116 \text{ nats}.
}
$$

---

### Percentage of uncertainty removed

The uncertainty reduction ratio is

$$
1-\frac{h(X\mid Y)}{h(X)}.
$$

Numerically,

$$
1-\frac{1.3073}{1.4189}
\approx 0.0787.
$$

So approximately

$$
\boxed{
7.9\%
}
$$

of the uncertainty about temperature is removed by observing the power meter.

---

### If the noise variance becomes 100

Suppose

$$
Q\sim\mathscr N(0,100).
$$

Then

$$
\mathrm{Var}(Y)=1+100=101.
$$

The conditional variance becomes

$$
\sigma_{X\mid Y}^2
=
1-\frac{1^2}{101}
=
\frac{100}{101}.
$$

Hence

$$
I(X;Y)
=
\frac12\ln\left(\frac{101}{100}\right).
$$

This is extremely small.

Therefore, if mechanical noise dominates the signal, the power measurement contains almost no useful information about temperature.

## 7. Diagnostic Interpretation

The manager says:

> "Since $Y=X+Q$, our best estimate should just be $Y$."

However this is incorrect because $Y$ contains both:

* true thermal signal,
* random mechanical noise.

The optimal estimator is

$$
E[X\mid Y]
=
\frac15Y.
$$

Thus the estimate is only one-fifth of the raw measurement.

This phenomenon is called **shrinkage toward the mean**.

Large observed values of $Y$ are not trusted completely because part of the observation may simply be noise from cutting torque.

For example, if

$$
Y=10,
$$

the best estimate is not

$$
X=10,
$$

but rather

$$
E[X\mid Y=10]=2.
$$

So the estimator discounts the observation according to the signal-to-noise ratio.

When the noise variance is large, the estimator shrinks even more strongly toward zero.
